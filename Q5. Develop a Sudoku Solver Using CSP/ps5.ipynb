```python

def is_valid(grid, row, col, num):
    # Check row
    for c in range(9):
        if grid[row][c] == num:
            return False

    # Check column
    for r in range(9):
        if grid[r][col] == num:
            return False

    # Check 3 x 3 subgrid
    start_row = (row // 3) * 3
    start_col = (col // 3) * 3

    for r in range(start_row, start_row + 3):
        for c in range(start_col, start_col + 3):
            if grid[r][c] == num:
                return False

    return True


def solve_sudoku(grid):
    # Find an empty cell (represented by 0)
    for row in range(9):
        for col in range(9):
            if grid[row][col] == 0:

                # Try digits 1 to 9
                for num in range(1, 10):
                    if is_valid(grid, row, col, num):
                        grid[row][col] = num

                        if solve_sudoku(grid):
                            return True

                        # Backtrack
                        grid[row][col] = 0

                return False

    return True


def validate_solution(grid):
    required = set(range(1, 10))

    # Check rows
    for row in grid:
        if set(row) != required:
            return False

    # Check columns
    for col in range(9):
        if {grid[row][col] for row in range(9)} != required:
            return False

    # Check 3 x 3 subgrids
    for start_row in range(0, 9, 3):
        for start_col in range(0, 9, 3):
            values = set()

            for r in range(start_row, start_row + 3):
                for c in range(start_col, start_col + 3):
                    values.add(grid[r][c])

            if values != required:
                return False

    return True


# Example Sudoku puzzle (0 means empty cell)
grid = [
    [5, 3, 0, 0, 7, 0, 0, 0, 0],
    [6, 0, 0, 1, 9, 5, 0, 0, 0],
    [0, 9, 8, 0, 0, 0, 0, 6, 0],
    [8, 0, 0, 0, 6, 0, 0, 0, 3],
    [4, 0, 0, 8, 0, 3, 0, 0, 1],
    [7, 0, 0, 0, 2, 0, 0, 0, 6],
    [0, 6, 0, 0, 0, 0, 2, 8, 0],
    [0, 0, 0, 4, 1, 9, 0, 0, 5],
    [0, 0, 0, 0, 8, 0, 0, 7, 9]
]

print("Original Sudoku:")
for row in grid:
    print(row)

if solve_sudoku(grid):
    print("\nCompleted Sudoku:")
    for row in grid:
        print(row)

    if validate_solution(grid):
        print("\nValidation successful! Sudoku solution is valid.")
    else:
        print("\nSolution failed validation.")
else:
    print("\nNo solution exists for this Sudoku puzzle.")
```
