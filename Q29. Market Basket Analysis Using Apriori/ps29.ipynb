```python
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules

#used to find items that frequently occur together in a dataset.
#For example, if {Milk, Eggs} does not meet the minimum support, then {Milk, Eggs, Bread} can be eliminated without checking its support.

# Step 1: Create a transactional dataset (10 transactions)
transactions = [
    ["Milk", "Bread", "Butter"],
    ["Bread", "Butter"],
    ["Milk", "Bread"],
    ["Milk", "Bread", "Butter", "Eggs"],
    ["Bread", "Eggs"],
    ["Milk", "Butter"],
    ["Milk", "Bread", "Eggs"],
    ["Bread", "Butter", "Eggs"],
    ["Milk", "Bread", "Butter"],
    ["Milk", "Eggs"]
]

print("Total Transactions:", len(transactions))

# Step 2: Convert transactions into one-hot encoded format
encoder = TransactionEncoder()
encoded_array = encoder.fit_transform(transactions) #Boolean transaction matrix

df = pd.DataFrame(
    encoded_array,
    columns=encoder.columns_
)

print("\nOne-Hot Encoded Dataset:")
print(df)

# Step 3: Apply Apriori with minimum support = 0.3 -> itemset must appear in at least (0.3*10)3 transactions to be considered frequent.
frequent_itemsets = apriori(
    df,
    min_support=0.3,
    use_colnames=True
)

print("\nFrequent Itemsets:")
print(frequent_itemsets)

# Step 4: Generate association rules
rules = association_rules(
    frequent_itemsets,
    metric="confidence",  #confidence -> Suppose 7 transactions contain Milk, and 5 of those contain both Milk and Bread.
    min_threshold=0.6
)

# Step 5: Display support, confidence and lift
def convert_items(x):
    return ", ".join(sorted(x))

rules["antecedents"] = rules["antecedents"].apply(convert_items)  #["Bread", "Milk"] -> Bread, Milk
rules["consequents"] = rules["consequents"].apply(convert_items)

print("\nAssociation Rules:")
print(
    rules[[
        "antecedents",  # the condition or starting item.
        "consequents",  #item associated with that condition
        "support",     #tansaction x/ total transaction (proportion)
        "confidence",
        "lift"          # confidence/support  (1< positive, 1> negative association, 1->independent)
    ]].round(3).to_string(index=False)
)
```
